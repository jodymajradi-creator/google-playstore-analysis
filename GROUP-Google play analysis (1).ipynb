Data Prepartion:

```python
import pandas as pd

df = pd.read_csv('googleplaystore.csv')
print('Shape:', df.shape)
df.head()
```

```python
import re
import numpy as np
import pandas as pd

df_clean = df.copy()

df_clean = df_clean[df_clean['Category'] != '1.9']
df_clean.drop_duplicates(inplace=True)

df_clean['Installs'] = df_clean['Installs'].astype(str).str.replace(r'[+,]', '', regex=True)
df_clean['Installs'] = pd.to_numeric(df_clean['Installs'], errors='coerce')

df_clean['Price'] = df_clean['Price'].astype(str).str.replace(r'[$]', '', regex=True)
df_clean['Price'] = pd.to_numeric(df_clean['Price'], errors='coerce')

def clean_size(val):
    if pd.isna(val) or str(val).strip() == 'Varies with device':
        return np.nan
    s = str(val).strip()
    if 'M' in s:
        return float(re.sub(r'[M]', '', s))
    elif 'k' in s or 'K' in s:
        return float(re.sub(r'[kK]', '', s)) / 1024.0
    return pd.to_numeric(s, errors='coerce')

df_clean['Size'] = df_clean['Size'].apply(clean_size)

df_clean['Reviews'] = pd.to_numeric(df_clean['Reviews'], errors='coerce')

df_clean['Last Updated'] = pd.to_datetime(df_clean['Last Updated'], errors='coerce')

df_clean['Rating'] = df_clean['Rating'].fillna(df_clean['Rating'].median())
df_clean['Size'] = df_clean['Size'].fillna(df_clean['Size'].median())
df_clean['Installs'] = df_clean['Installs'].fillna(df_clean['Installs'].median())
df_clean['Price'] = df_clean['Price'].fillna(df_clean['Price'].median())
df_clean['Reviews'] = df_clean['Reviews'].fillna(df_clean['Reviews'].median())

df_clean.dropna(subset=['Type', 'Content Rating', 'Current Ver', 'Android Ver', 'Last Updated'], inplace=True)

print(df_clean.shape)
print(df_clean.dtypes)
print(df_clean.isnull().sum())
```

```python
import os
import pandas as pd

apps_file = "googleplaystore.csv" if os.path.exists("googleplaystore.csv") else "googleplaystore"
reviews_file = "googleplaystore_user_reviews.csv" if os.path.exists("googleplaystore_user_reviews.csv") else "googleplaystore_user_reviews"

apps = pd.read_csv(apps_file)
reviews = pd.read_csv(reviews_file)

apps = apps.drop_duplicates(subset=["App"]).copy()
apps["Rating"] = pd.to_numeric(apps["Rating"], errors="coerce")
apps["Installs_num"] = (
    apps["Installs"]
    .astype(str)
    .str.replace(r"[+,]", "", regex=True)
    .str.strip()
)
apps["Installs_num"] = pd.to_numeric(apps["Installs_num"], errors="coerce")

reviews_summary = reviews.groupby("App", as_index=False).size().rename(columns={"size": "reviews_count"})
merged_df = pd.merge(apps, reviews_summary, on="App", how="left")

summary_by_category = (
    merged_df.groupby("Category", as_index=False)
    .agg(
        avg_rating=("Rating", "mean"),
        total_installs=("Installs_num", "sum"),
        apps_count=("App", "count")
    )
    .sort_values(by="total_installs", ascending=False)
)

display(summary_by_category.head(10))

merged_df.to_csv("clean_playstore_dataset.csv", index=False)
```
